```python
import pandas as pd

df_normal = pd.read_csv("dataset_정상.csv", encoding='cp949', dtype={'업체코드': str})
df_normal.columns
```

```python
import pandas as pd

df_failed = pd.read_csv("dataset_부실.csv", encoding='cp949', dtype={'업체코드': str})

df_failed.columns
```

```python
drop_cols = ['코스피구분', '상폐사유', '상폐사유요약']

df_failed = df_failed.drop(columns=drop_cols, errors='ignore')

# 3. 결과 확인 (제거된 후 컬럼 목록 출력)
print(f"✅ 컬럼 제거 완료! 현재 남은 컬럼 수: {len(df_failed.columns)}")
print(f"남은 주요 컬럼 예시: {df_failed.columns.tolist()}")
```

```python
import pandas as pd


print(f"정상 데이터 행/열: {df_normal.shape}")
print(f"부실 데이터 행/열: {df_failed.shape}")

# 2. 컬럼 순서 일치시키기 (매우 중요)
# 컬럼 이름이 같더라도 순서가 다르면 데이터가 꼬일 수 있으므로 정렬합니다.
common_cols = sorted(df_normal.columns.tolist())
df_normal = df_normal[common_cols]
df_failed = df_failed[common_cols]

# 3. 데이터 수직 통합
# ignore_index=True를 주어 인덱스를 0부터 다시 새로 매깁니다.
df_final = pd.concat([df_normal, df_failed], axis=0, ignore_index=True)

# 4. 정렬 및 저장
# 업체코드와 연도별로 보기 좋게 정렬
df_final = df_final.sort_values(by=['업체코드', '연도']).reset_index(drop=True)

df_final.to_csv("dataset.csv", index=False, encoding='utf-8-sig')

print("\n🚀 라벨링 없이 단순 통합 완료!")
print(f"최종 데이터 크기: {df_final.shape}")
```
